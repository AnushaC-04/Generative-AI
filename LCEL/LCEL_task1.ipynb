```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
llm=ChatGoogleGenerativeAI(
    model='gemini-3.5-flash',
    api_key=os.environ['GEMINI_API_KEY'],
        temperature=0.2
    )
```

```python
lang_detect_prompt_template=ChatPromptTemplate.from_messages([
    ('system',
     "You're a helpful assistant."
     "Detect the native from the given text"
     "Return only Detected language"),
     ('human',"{text}")
])
```

```python
parser=StrOutputParser()
```

```python
language_chain=lang_detect_prompt_template |llm | parser
# language=language.invoke({"prompt":"Machine learning oru easyana subject,ithala naraya algorithms iruku ellamey easy ah irukum"})
```

### LCEL -2 --> for translation

```python
translate_prompt_template=ChatPromptTemplate.from_messages([
    ('system',
     "You're a helpful assistant."
     "Detect the language from the given text and translate the given text to English Language"
     "Return the text and the Translated text"),
     ('human',"{text}")
])
```

```python
translate_chain=translate_prompt_template |llm | parser
# translate_text=translate_text.invoke({"prompt":"Machine learning oru easyana subject,ithala naraya algorithms iruku ellamey easy ah irukum"})
```

```python
# translate_text
```

### LCEL-3 --> for summarizing

```python
summarize_prompt_template=ChatPromptTemplate.from_messages([
    ('system',
     "You're a helpful assistant."
     "summarize the given text "
     "Return only the Summarized text"),
     ('human',"{text}")
])
```

```python
summarize_chain=summarize_prompt_template |llm | parser
# summarize=summarize.invoke({"text":translate_text})
```

```python
# summarize
```

```python
full_chain=  language_chain | translate_chain | summarize_chain
```

```python
final_response=full_chain.invoke({'text':"Machine learning oru easyana subject,ithala naraya algorithms iruku ellamey easy ah irukum"})
```
