```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
import os
```

```python
llm=ChatGoogleGenerativeAI(
    model='gemini-3.5-flash',
    api_key=os.environ['GEMINI_API_KEY'],
    temperature=0.0
)
```

```python
parser=StrOutputParser()
```

```python
extract_keyword_prompt=ChatPromptTemplate(
    messages=[
    (
        "system",
        """You are a helpful assistant.
        Extract the important technical keywords from the given text.
        Return only the keywords separated by commas."""
    ),
        ("human", "Text: {text}")
])
```

```python
keyword_chain=extract_keyword_prompt | llm | parser
```

```python
categorize_prompt=ChatPromptTemplate(
    messages=[
        ("system",
                "You are a helpful assistant"
                "for the given keyword {text} categorize them into their appropriate technical categories."
                "return each keyword with its appropriate technical category"
                "Example:"
                "Python → Programming"
                "Pandas → Data Analysis"
                "NumPy → Numerical Computing"
                "Scikit-learn → Machine Learning"
                ),
        ("human","keywords: {text}")
    ]
)
```

```python
categorize_chain=categorize_prompt | llm | parser
```

```python
full_chain = keyword_chain | categorize_chain
```

```python
full_response=full_chain.invoke({'text':"""Python is widely used in data analysis and machine learning.
Pandas and NumPy are useful for data manipulation and numerical
computations. Matplotlib and Seaborn are commonly used for data
visualization, while Scikit-learn provides machine learning algorithms
such as linear regression, decision trees, and clustering."""})
```
