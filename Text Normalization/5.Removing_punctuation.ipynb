```python
with open('data.txt','r+') as file:
    data=file.read()
    punctuations=string.punctuation
    for i in punctuations:
        data=data.replace(i,'')
        file.write(data)
        file.seek(0)  
print(data)
```

```python
'''Alternative way-1 '''
with open('data.txt','r') as file:
    data=file.read()
    for char in data:
        if not char.isalnum() and not ord(char)==32:
            data=data.replace(char,'')
            # print(data)    # this print stmt is used to debug the program
print(data)
```

```python
'''Alternative way-2'''
import re
data=re.sub(r'[^0-9a-zA-Z\s]','',data)
```
