```python
with open('data.txt','r+') as file:
    data=file.read()
    data=data.lower()
    file.seek(0)
    file.write(data)
print(data)
```

### Converting the string to lower case

```python
with open('data.txt','r') as file:
    data=file.read().lower()
print(data)
```

### 2. removing punctuation

```python

import string 
result=string.punctuation
result
```

```python
# method -1 : IN existing string

with open('data.txt','r') as file:
    data=file.read()
    for char in data:
        if char in result:
            data=data.replace(char , '')
print(data)
```
