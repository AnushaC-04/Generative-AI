```python
'''
Upload Document
      │
      ▼
Read Document
      │
      ▼
Convert to Lowercase
      │
      ▼
Remove Extra Spaces
      │
      ▼
Handle Emojis
      │
      ├── Option 1: Remove Emojis
      └── Option 2: Convert Emojis to Text
      │
      ▼
Remove Punctuation
      │
      ▼
Word Tokenization
      │
      ▼
Convert Tokens to Paragraph
      │
      ▼
TextBlob
      │
      ▼
spaCy Lemmatization
      │
      ▼
Recursive Chunking
'''
```

```python
import string
import emoji
import spacy
from nltk.tokenize import word_tokenize
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Read document & lowercase
with open('data.txt', 'r', encoding='utf-8') as file:
    data = file.read().lower()

# 2. FIXED ORDER: Remove emojis FIRST before changing unicode structure
data = emoji.replace_emoji(data, '')

# 3. Remove punctuation
for char in string.punctuation:
    data = data.replace(char, '')

# 4. FIXED PERFORMANCE: Filter special chars without re-creating strings in a loop
data = "".join([char for char in data if char.isalnum() or char == " "])

# 5. Remove extra spaces
data = " ".join(data.split())

# 6. NLTK Tokenization -> Re-join
word_tokens = word_tokenize(data, language="english")
paragraph = " ".join(word_tokens)

# 7. spaCy Lemmatization
nlp = spacy.load('en_core_web_sm', disable=["parser", "ner"])
doc = nlp(paragraph)

lemmatized_tokens = [token.lemma_ for token in doc if not token.is_space]
lemmatized_paragraph = " ".join(lemmatized_tokens)

# 8. LangChain Chunking
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separators=["\n\n", "\n", " ", ""]
)

chunks = text_splitter.split_text(lemmatized_paragraph)

print(f"Successfully generated {len(chunks)} chunks!")
print(chunks)
```
