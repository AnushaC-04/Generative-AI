```python
from nltk.stem import PorterStemmer

stemmer=PorterStemmer()
```

```python
words = [
    "running",
    "playing",
    "studies",
    "studying",
    "happily",
    "better",
    "walking",
    "walked",
    "jumps",
    "jumping",
    "cats",
    "children",
    "flying",
    "flies",
    "easily",
    "writing",
    "written",
    "driving",
    "driver",
    "connection",'a','the'
]
```

- stemming works based on rules it doesnt convert to root form just remove the suffix

- it does 2 things 
  - it remove the suffix based on rules
  - it convert the string to lowercase

### solution to convert words to its root form using **lamendization**

```python
for word in words:
    print(f'{word} --> {stemmer.stem(word)}')
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball=SnowballStemmer(language='english',ignore_stopwords=True)
```

```python
words = [
    "running",
    "runner",
    "runs",
    "easily",
    "fairly",
    "studying",
    "studied",
    "studies",
    "playing",
    "played",
    "happiness",
    "happily",
    "organization",
    "organizing",
    "connection",
    "connected",
    "relational",
    "conditional",
    "beautifully",
    "information" ,'a','the','between'
]
```

```python
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
stop_words=stopwords.words('english')
# stop_words

snowball_stemmer=SnowballStemmer("english",ignore_stopwords=True)
snowball_stemmer.stop_words=stop_words
snowball_stemmer.stop_words
for word in words:
    print(f'{word} --> {snowball_stemmer.stem(word)}')
```
