```python
import nltk
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')  #pos 
nltk.download('averaged_perceptron_tagger_eng')
```

```python
Lemmatizer=WordNetLemmatizer()
```

```python
tokens = [
    "running",
    "cars",
    "children",
    "better",
    "wolves",
    "studies",
    "playing",
    "mice",
    "geese",
    "feet"
]
```

```python
for word in tokens:
    print(word, Lemmatizer.lemmatize(word))   #default pos is noun-->'n'
```

```python
for word in tokens:
    print(word, Lemmatizer.lemmatize(word,pos='n'))  #if it the noun then it apply lementize
```

```python
for word in tokens:
    print(word, Lemmatizer.lemmatize(word,pos='v')) #pos- part of speech
```

```python
nltk.pos_tag(tokens)

'''
The Part Of Speech tag. 
 "NNS"-->"n" for nouns
 "VBG"-->"v" for verbs
 "JJ"--> "a" for adjective
 "RBR"-->"r" for adverbs
 "NNS"-->"s" for satellite adjectives.

remaining all we will give it as noun
'''
```

```python
def find_postag(tag):
    if tag.startswith('J'):
        return 'a'
    elif tag.startswith('N'):
        return 'n'
    elif tag.startswith('V'):
        return 'v'
    elif tag.startswith('R'):
        return 'r'
    else:
        return 'n'
```

```python
pos_tag_words=nltk.pos_tag(tokens)
pos_tag_words
```

```python
for word,tag in pos_tag_words: #('running', 'VBG')
    tagged_words=Lemmatizer.lemmatize(word,find_postag(tag))
     # Lemmatizer.lemmatize('running',v)
    print(tagged_words)

#manually we need to find the part of speech which is the drawback
```
