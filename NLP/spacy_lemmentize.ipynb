```python
import spacy
nlp=spacy.load('en_core_web_sm')

tokens=nlp(''' Apply the pipeline to some Text. If True, a vocab is created. disable Union[str, Iterable[str]. Name(s) of pipeline component(s) to disable.''')
print(type(tokens),tokens)
```

- we can perform :
    - tokenization
    - pos tag
    - lementization
    - name entity recongnition

- first by default it convert this into tokens ,thus tokenization is not required
- second after converting to lokens it by default it convert each the token to lower case
- then it apply post_tagging by default (it recognize the words and then convert it into its root form ) 

```python
lemmatize_tokens= [token.lemma_ for token in tokens if not token.is_stop]
print(lemmatize_tokens)
print(' '.join(lemmatize_tokens))

# for token in tokens:
#     print(token.lemma_,token.is_stop)
# print(type(token))
```

```python
for token in tokens:
    print(token.pos_,token.tag_)
```
